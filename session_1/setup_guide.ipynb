## Part 2: Get Your Gemini API Key

### What is an API Key?

An API key is like a password that lets your code talk to Google's AI. It's a long string of letters and numbers that looks like: `AIzaSyD...`

### Steps to Get Your Key

1. **Go to Google AI Studio**: [aistudio.google.com/apikey](https://aistudio.google.com/apikey)
2. **Sign in** with your Google account (same as Gmail)
3. **Click "Create API Key"**
4. **Enable billing when prompted** — Google requires a billing method on file before it will issue a key (see the note below)
5. **Let it create a new project** (just click "Continue" when asked)
6. **Copy the key** that appears (starts with `AIza`)

**Total time**: ~2 minutes

### ⚠️ Billing must be enabled

As of 2026, Google requires a billing method on file before it will issue a Gemini API key at all — this applies everywhere now, not just the EU/UK/Switzerland (which required it earlier, for regulatory reasons).

Don't worry about a surprise bill: the lightweight models this course uses cost fractions of a cent per request, and a typical session here uses well under a dollar's worth of usage. Still, there's no free quota to fall back on if you run something large by mistake — check the token cost guide before a big batch job, and keep an eye on your usage at the [AI Studio dashboard](https://aistudio.google.com).

### Storing Your API Key Securely in Colab

**DO NOT paste your API key directly in code!** Here's the secure way:

1. Look at the left sidebar in Colab
2. Click the **🔑 key icon** (called "Secrets")
3. Click **"+ Add new secret"**
4. **Name**: `GEMINI_API_KEY`
5. **Value**: Paste your API key (the one starting with `AIza`)
6. Toggle the **switch** to allow this notebook to access it
7. Click **"Save"**

Now let's test if it worked:

````python
# Check the key is stored and readable
try:
    from google.colab import userdata
    api_key = userdata.get('GEMINI_API_KEY')
    print("Key loaded. Starts with:", api_key[:10] + "...")
except Exception as e:
    print("Could not load the key.")
    print("Check it is saved in Secrets as GEMINI_API_KEY and the access toggle is on.")
    print("Error:", e)
````

## Part 5: your usage and rate limits

Billed usage still has per-minute and per-day rate limits — paying doesn't mean unlimited. Google changes these limits periodically and they vary by model, so check your live limits in [AI Studio](https://aistudio.google.com) rather than relying on a number quoted here.

A typical session in this course uses roughly 20 to 50 requests, which costs a small fraction of a cent on the lightweight models we use. If you hit a rate limit, the API returns a 429 error — wait for the reset, or check whether your account needs a higher tier.

Estimating token costs before running a job is covered in this week's cost guide — worth doing every time, since there's no free quota to absorb a mistake.

The counter below tracks requests made in this notebook session only, and resets when the notebook restarts. Call `count_request()` after each API call you want to track.

````python
request_count = 0

def count_request():
    global request_count
    request_count += 1
    print("Requests this session:", request_count)

count_request()
````

### Sentiment analysis on a customer review

A classification task, in the sense covered in Lesson 6. Note the review below is deliberately mixed, which is where sentiment models are least reliable.

````python
from google import genai
from google.colab import userdata

client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))

review = "The product arrived late, but the quality is amazing. Worth the wait!"

prompt = f'''Analyze the sentiment of this customer review.
Review: "{review}"

Provide:
1. Overall sentiment (positive/negative/mixed)
2. Confidence score (0-100%)
3. Key phrases
'''

response = client.models.generate_content(
    model='gemini-2.5-flash-lite',
    contents=prompt
)

print(response.text)
````

### Generating marketing copy from a brief

````python
from google import genai
from google.colab import userdata

client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))

product = "eco-friendly water bottle"
target_audience = "environmentally conscious millennials"

prompt = f'''Write 3 short social media captions (max 280 characters each) for:
Product: {product}
Target audience: {target_audience}

Make them engaging and include a call-to-action.
'''

response = client.models.generate_content(
    model='gemini-2.5-flash-lite',
    contents=prompt
)

print(response.text)
````

Both prompts describe the task without including worked examples, which makes them zero-shot (Lesson 6). Adding two or three labelled examples is the first thing to try when output is inconsistent.

### Installation steps

This course uses [uv](https://docs.astral.sh/uv/) for local setup rather than a hand-built virtual environment, since it gives each session its own fresh, isolated environment automatically, with no `venv` to create or activate by hand.

```bash
# Clone the repository (one time)
git clone https://github.com/YOUR_USERNAME/ai4tm-masterclass.git
cd ai4tm-masterclass

# Install uv (one time)
curl -LsSf https://astral.sh/uv/install.sh | sh
# Windows (PowerShell): powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

From here, pick whichever of the two matches how you like to work. Both end up running the exact same notebook cells the exact same way; only how you open the notebook differs.

**Option 1: a browser tab, like Colab.** Each time you want to work on a session, `cd` into that session's folder and run:

```bash
cd session_1
uv run --with jupyter jupyter lab
```

This opens Jupyter Lab in your browser, with a fresh environment built just for that session. Open the notebook from the tab that appears and run cells as normal. Every notebook's own `!pip install` cells still work exactly the same way inside this environment.

**Option 2: inside VS Code.** If you'd rather stay in an editor than switch to a browser tab, create a persistent environment once per session folder and point VS Code at it instead:

```bash
cd session_1
uv venv
uv pip install ipykernel
```

Then open the notebook file in VS Code, click **Select Kernel** in the top right, choose **Python Environments**, and pick the `.venv` that was just created. Everything runs inline in VS Code from there, no browser tab involved. Run the notebook's own setup cells as normal once that kernel is selected. They install the actual libraries (`google-genai` and so on) into whichever environment you picked, so this step and that one are separate: this gets you a working kernel, the notebook's setup cell is what fills it with the libraries a given session needs.

### Storing API keys locally

Create a file called `.env` in your project folder (after `pip install python-dotenv` inside whichever environment you set up above):

```
GEMINI_API_KEY=your_key_here
```

Load it in Python:

```python
from dotenv import load_dotenv
import os

load_dotenv()
api_key = os.getenv('GEMINI_API_KEY')
```

**Important**: The `.env` file is already in `.gitignore` so it won't be uploaded to GitHub.
